```python
### esto es solo para tener claro cómo se introduce la api en python
import pandas as pd
import requests

url = "https://servicios.ine.es/wstempus/js/es/DATOS_TABLA/31097?tip=AM"


response = requests.get(url)
data = response.json()

df_rents = []
for serie in data:
    nombre_serie = serie['Nombre']
    for dato in serie['Data']:
        df_rents.append({
            'Serie': nombre_serie,
            'Fecha': dato['Fecha'],
            'Valor': dato['Valor']
        })

df_renta = pd.DataFrame(df_rents)

print(df_renta.head())
```

```python
import pandas as pd
import requests

# URL de la tabla del INE
url = "https://www.ine.es/jaxiT3/Datos.htm?t=31097"

# El INE puede bloquear peticiones sin un User-Agent (simulamos un navegador)
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"
}

try:
    # 1. Realizar la descarga del HTML
    response = requests.get(url, headers=headers)
    response.encoding = 'utf-8' # Aseguramos tildes y caracteres especiales

    # 2. Leer las tablas del HTML
    # Configuramos miles y decimales para obtener tipos numéricos (float/int)
    tablas = pd.read_html(response.text, decimal=',', thousands='.')

    # 3. Seleccionar la tabla principal (suele ser la primera)
    df = tablas[0]

    # 4. Limpieza básica: Eliminamos filas de totales si no las necesitas
    # y reseteamos el índice
    df = df.dropna(how='all').reset_index(drop=True)

    print("¡Tabla extraída con éxito!")
    print(df.head())

except Exception as e:
    print(f"Error al extraer los datos: {e}")
```
